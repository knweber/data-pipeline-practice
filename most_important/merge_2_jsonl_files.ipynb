```python
import json
```

```python
customer_lookup = {}

with open("../sample_data/sample_customers.jsonl", "r", encoding="utf-8") as cust_file:
    for line in cust_file:
        cust = json.loads(line)
        customer_lookup[cust["customer_id"]] = cust

print(customer_lookup)
```

```python
with open("../sample_data/sample_orders.jsonl", "r", encoding="utf-8") as order_file, open("../output/merged_jsonl.jsonl", "w", encoding="utf-8") as outfile:
    for line in order_file:
        order = json.loads(line)
        cust_id = order.get("customer_id")

        if cust_id in customer_lookup:
            for key, val in customer_lookup[cust_id].items():
                if key != "customer_id":
                    order[f"customer_{key}"] = val

        outfile.write(json.dumps(order) + "\n")
```
