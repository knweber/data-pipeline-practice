# writing csv

```python
# read data
import csv

customer_lookup = {}

with open("../sample_data/sample_customers.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    headers = reader.fieldnames
    for line in reader:
        customer_lookup[line["customer_id"]] = line
```

```python
# write data
with open("../output/test.csv", "w", newline="", encoding="utf-8") as outfile:
    writer = csv.DictWriter(outfile, fieldnames = headers)
    writer.writeheader()

    for outer_key, inner_dict in customer_lookup.items():
        row = {'customer_id': outer_key}
        row.update(inner_dict)
        writer.writerow(row)
```

# writing json

```python
# read data
import json

records = []

with open("../sample_data/sample_customers.json", "r", encoding="utf-8") as file:
    data = json.load(file)
    for row in data:
        records.append(row)

print(records)
```

```python
# write data
with open("../output/test.json", "w", encoding="utf-8") as file:
    #file.write(json.dumps(records))
    json.dump(records, file, indent=4)
```

# writing jsonl

```python
# read data
import json

records = []

with open("../sample_data/sample_customers.jsonl", "r", encoding="utf-8") as file:
    for line in file:
        cust = json.loads(line)
        records.append(cust)

print(records)
```

```python
# write data
with open("../output/test.jsonl", "w", encoding="utf-8", newline="") as file:
    for record in records:
        file.write(json.dumps(record) + "\n")
```
