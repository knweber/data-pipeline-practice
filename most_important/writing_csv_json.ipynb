# writing csv

```python
# read data
import csv
from datetime import datetime

customers = []

with open("../sample_data/sample_customers.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for line in reader:
        customers.append(line)
```

```python
# write data
transformed_records = []
curr_time = datetime.now()

for cust in customers:
    transformed = {
        "id": cust.get("customer_id"),
        "name": cust.get("customer_name"),
        "processed_at": datetime.strftime(curr_time, '%Y-%m-%dT%H:%M:%S')
    }
    transformed_records.append(transformed)
    headers = transformed.keys()


with open("../output/test.csv", "w", newline="", encoding="utf-8") as outfile:
    writer = csv.DictWriter(outfile, fieldnames = headers)
    writer.writeheader()
    for rec in transformed_records:
        writer.writerow(rec)
```

# writing json

```python
# read data
import json

records = []

with open("../sample_data/sample_customers.json", "r", encoding="utf-8") as file:
    data = json.load(file)
    for row in data:
        records.append(row)

print(records)
```

```python
# write data
with open("../output/test.json", "w", encoding="utf-8") as file:
    #file.write(json.dumps(records))
    json.dump(records, file, indent=4)
```

# writing jsonl

```python
# read data
import json

records = []

with open("../sample_data/sample_customers.jsonl", "r", encoding="utf-8") as file:
    for line in file:
        cust = json.loads(line)
        records.append(cust)

print(records)
```

```python
# write data
with open("../output/test.jsonl", "w", encoding="utf-8", newline="") as file:
    for record in records:
        file.write(json.dumps(record) + "\n")
```
