```python
import sqlite3
import requests

URL = "https://typicode.com"
DB_NAME = "test_db.db"
```

```python
def extract_data(url):
    try:
        response = requests.get(url)
        response.raise_for_status()
        return response.json()
    except Exception as e:
        print(f"Exception occurred: {e}")
    
```

```python
def transform_data(raw_users):
    transformed_users = []
    for user in raw_users:
        clean_user = {
            "id": user.get("id"),
            "name": user.get("name", "").strip().upper(),
            "email": user.get("email", "").lower(),
            "city": user.get("address", {}).get("city"),
            "company": user.get("company",{}).get("name"),
        }
        transformed_users.append(clean_user)
    return transformed_users
```

```python
def load_data(cleaned_users, db_name):
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()

    cursor.execute("""
    CREATE TABLE IF NOT EXISTS users (
        id INTEGER PRIMARY KEY,
        name TEXT,
        email TEXT,
        city TEXT,
        company TEXT
    )
    """)

    insert_query = """
        INSERT INTO users (id, name, email, city, company) VALUES (:id, :name, :email, :city, :company)
    """

    cursor.executemany(insert_query, cleaned_users)

    data = cursor.execute('''SELECT * FROM users''')
    for item in data:
        print(item)

    conn.commit()
    conn.close()
    print(f"Successfully loaded {len(cleaned_users)} records")
```

```python
def run_pipeline():
    try:
        print("Extracting data ...")
        raw_data = extract_data(URL)

        print("Transforming data ...")
        cleaned_data = transform_data(raw_data)

        print("Loading data ...")
        load_data(cleaned_data, DB_NAME)

        print("ETL job finished successfully")

    except requests.exceptions.RequestException as e:
        print(f"Extraction failed: {e}")
    except sqlite3.Error as e:
        print(f"Database operation failed: {e}")
```

```python
run_pipeline()
```

```python
import sqlite3
import requests

API_URL = "https://rickandmortyapi.com/api/character"
DB_NAME = "test.db"
```

```python
def extract(url):
    print("Extracting ...")
    try:
        response = requests.get(url)
        response.raise_for_status()
        print("Extraction successful")
        return response.json()
    except requests.exceptions.Exception as e:
        print(f"Exception occcurred during API call: {e}")
    except Exception as e:
        print(f"Exception occurred: {e}")
```

```python
def transform(data):
    print("Transforming ...")
    #print(data)
    transformed_data = []
    for char in data['results']:
        transformed = {
            "id": char.get("id"),
            "name": char.get("name")
        }
        transformed_data.append(transformed)
    print("Transformation successful")
    #print(transformed_data)
    return transformed_data
```

```python
def load(cleaned_users, db_name):
    print("Loading to DB")
    print("Cleaned users: ")
    print(cleaned_users)
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS rick_morty (
            id INTEGER PRIMARY KEY,
            name TEXT
        )
    """)

    insert_query = """
        INSERT INTO rick_morty (id, name) VALUES (:id, :name)
    """

    cursor.executemany(insert_query, cleaned_users)
    data = cursor.execute('''SELECT * FROM rick_morty''')
    for item in data:
        print(f"*** {item}")

    conn.commit()
    conn.close()
    print(f"Successfully loaded {len(cleaned_users)} records")
```
