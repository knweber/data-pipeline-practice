```python
import json
```

```python
with open("../sample_data/sample_customers.json", "r", encoding="utf-8") as infile:
    data = json.load(infile)

with open("../output/jsonl.jsonl", "w", encoding="utf-8") as outfile:
    for item in data:
        outfile.write(json.dumps(item) + "\n")

print("Conversion complete")
```
