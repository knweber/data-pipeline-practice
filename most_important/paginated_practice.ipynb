```python
import requests
import json
from datetime import datetime
import time

URL = "https://rickandmortyapi.com/api/character"
OUTPUT_FILE = "../output/rick_morty2.jsonl"
```

```python
def extract_data(start_url):
    next_url = start_url
    page_count = 1
    max_page_count = 10
    all_data = []

    while next_url and page_count <= max_page_count:
        print(f"Extracting page {page_count}")
        try:
            response = requests.get(next_url)
            response.raise_for_status()

            if response.status_code == 429:
                retry_after = int(response.headers.get("Retry-After"))
                time.sleep(retry_after)
                continue

            payload = response.json()

        except json.JSONDecodeError as e:
            print(f"JSON Decode Error: {e}")
        except Exception as e:
            print(f"Exception occcurred: {e}")

        data = payload.get("results", [])
        print(data)

        all_data.extend(data)
        next_url = payload.get("info").get("next")
        page_count += 1

    return all_data
```

```python
extract_data(URL)
```

```python
def transform(char):
    curr_time = datetime.now()

    return {
        "character_id": char.get("id"),
        "name": char.get("name"),
        "gender": char.get("gender"),
        "origin_name": char.get("origin").get("name"),
        "human": True if char.get("species") == "Human" else False,
        "extracted_at": curr_time.strftime("%Y-%m-%dT%H:%M:%S")
    }
```

```python
def run_pipeline(url, output_file):
    record_count = 0
    data = extract_data(url)
    with open(output_file, "w", encoding="utf-8") as f:
        for record in data:
            char = transform(record)
            print(char)
            f.write(json.dumps(char) + "\n")
            record_count += 1

    print(f"Wrote {record_count} records")
```

```python
run_pipeline(URL, "../output/test_json.jsonl")
```
