```python
events = [{"trip_id": "t1", "event": "started", "timestamp": "2026-08-01T10:00:00", "source": "mobile"}, {"trip_id": "t1", "event": "not_started", "timestamp": "2026-08-01T10:00:00", "source": "mobile"}, {"trip_id": "t2", "event": "started", "timestamp": "2026-08-02T10:00:00", "source": "mobile"}, {"trip_id": "t2", "event": "started", "timestamp": "2026-08-03T10:00:00", "source": "mobile"}]
```

```python
def dedup_events(events):
    earliest = {}

    for e in events:
        key = (e["trip_id"], e["event"])
        print(f"Key: {key}")
        
        generated_ts = e.get("timestamp")

        if key not in earliest or generated_ts < earliest[key]["timestamp"]:
            earliest[key] = e

    print(earliest.values())
    return list(earliest.values())
```

```python
dedup_events(events)
```

```python
bookings = [
    {"booking_id": "b1", "host_id": "h1", "vehicle_id": "v1", "amount": 120.50, "status": "completed", "date": "2026-08-01"}, 
    {"booking_id": "b2", "host_id": "h1", "vehicle_id": "v1", "amount": 130, "status": "completed", "date": "2026-08-01"}, 
    {"booking_id": "b3", "host_id": "h2", "vehicle_id": "v2", "amount": 50, "status": "completed", "date": "2026-08-02"},
    {"booking_id": "b4", "host_id": "h3", "vehicle_id": "v2", "amount": 200, "status": "cancelled", "date": "2026-08-03"}
    ]
```

```python
from collections import defaultdict
from datetime import datetime

def agg(bookings):
    seen_booking_ids = set()
    revenue_host_day = defaultdict(float)

    for b in bookings:
        booking_id = b.get("booking_id")
        host = b.get("host_id")
        amount = b.get("amount")
        status = b.get("status")
        rental_date = b.get("date")

        if booking_id in seen_booking_ids:
            continue
        if status != "completed":
            continue

        seen_booking_ids.add(booking_id)
        revenue_host_day[(host, rental_date)] += amount

    print(dict(revenue_host_day))
```

```python
agg(bookings)
```
