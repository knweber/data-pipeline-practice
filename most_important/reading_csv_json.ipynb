# reading csv

```python
import csv

customer_lookup = {}

with open("../sample_data/sample_customers.csv", mode="r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    headers = reader.fieldnames
    print(headers)
    for line in reader:
        customer_lookup[line["customer_id"]] = line
        
print(customer_lookup)
```

# reading json

```python
import json

with open("../sample_data/sample_customers.json", mode="r", encoding="utf-8") as file:
    data = json.load(file)
    print(data)
```

# reading jsonl

```python
import json

customers = []

with open("../sample_data/sample_customers.jsonl", mode="r", encoding="utf-8") as file:
    for line in file:
        customer = json.loads(line)
        customers.append(customer)

print(customers)
```

# reading base64 encoded --> json

```python
import base64
import json

raw_base64_data = (
    "W3sibmFtZSI6ICJhbGljZSIsICJtYWlsIjogIkFMSUNFQGV4YW1wbGUuY29tIiwg"
    "ImFnZSI6IDMwfSwgeyJuYW1lIjogImJvYiIsICJtYWlsIjogIkJPQkVleGFtcGxl"
    "LmNvbSIsICJhZ2UiOiAyNX1d"
)

decoded_bytes = base64.b64decode(raw_base64_data) 
text_data = decoded_bytes.decode("utf-8")
records = json.loads(text_data)

print(records)
```
