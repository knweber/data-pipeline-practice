```python
import csv
import io
import json
from datetime import datetime

csv_data = """id,name,age
1,Bob,32
2,Cindy,25
3,Peter,41
4,Mildred,65
5,Beaux,4
6,Frank,72
"""
```

```python
transformed_data = []

string_file = io.StringIO(csv_data)
reader = csv.DictReader(string_file)

for row in reader:
    row["id"] = int(row["id"])
    row["age"] = int(row["age"])
    row["name"] = row["name"].lower()
    row["is_senior"] = row["age"] >= 60
    transformed_data.append(row)
```

```python
transformed_data = []
string_file = io.StringIO(csv_data)
reader = csv.DictReader(string_file)

extracted_at = datetime.now()

for row in reader:
    record = {
        "id": int(row.get("id")),
        "age": int(row.get("age")),
        "name": row.get("name").lower(),
        "is_senior": int(row.get("age")) >= 60,
        "extracted_at": datetime.strftime(extracted_at, "%Y-%m-%dT%H:%M:%S")
    }
    transformed_data.append(record)

print(transformed_data)
```

```python
print(transformed_data)
```

```python
headers = list(transformed_data[0].keys()) if transformed_data else []

print(headers)
```

```python
json_object = json.dumps(transformed_data)
print(json_object)
```

```python
transformed_data.sort(key=lambda x: x["age"])
```

```python
with open("../output/users_sorted_2.json", mode="w", encoding="utf-8") as f:
    json.dump(transformed_data, f, indent=4)

print("File successfully saved")
```

```python
with open("../output/users_sorted_2.jsonl", mode="w", encoding="utf-8") as f:
    for record in transformed_data:
        f.write(json.dumps(record) + "\n")

print("JSONL file written")
```

```python

with open("../output/users_sorted.csv", mode="w", newline='', encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=headers)
    writer.writeheader()
    writer.writerows(transformed_data)
```
