```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
import os
import sys

# 1. Force Python to find your Homebrew OpenJDK 11 path
os.environ["JAVA_HOME"] = "/opt/homebrew/opt/openjdk@17"

# 2. Tell PySpark exactly which Python executable to map to the JVM
os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

spark = SparkSession.builder \
    .master("local[*]") \
    .appName("ETL_Test") \
    .getOrCreate()

print("Spark Session Created Successfully!")
```

```python
#spark = SparkSession.builder.appName("ETLPipeline").getOrCreate()

df = spark.read.parquet("../sample_data/titanic.parquet")
print("Loaded from parquet file")
transformed_df = df.withColumn("Alive", F.when(F.col("Survived") == "1", "Yes").otherwise("No"))

print("Loading to parquet file")
transformed_df.write.mode("overwrite").parquet("../output/pyspark_titanic.parquet")
print("Succeeded")
```
